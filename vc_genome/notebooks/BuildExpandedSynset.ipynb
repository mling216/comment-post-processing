# Build Expanded Object Synset via Claude API

**Goal:** All object names in `oar_B_510.json` that fall through `_vc_canon.py`'s lookup
are returned as `unknown.<raw_name>`. This notebook:
1. Collects every such unmapped name (with corpus frequency)
2. Asks Claude Opus to assign each one to the existing `category.subcategory` taxonomy
3. Saves the result as **`scripts/_vc_canon_extended.py`** — a drop-in supplement that
   does **not** overwrite `_vc_canon.py`.

**Taxonomy categories** (inherited from `_vc_canon.py`):

| Category | Sub-categories |
|----------|----------------|
| `mark` | bar, point, line, color_segment, glyph, node, area, shape, symbol, element |
| `property` | color, font_size, encoding |
| `structure` | panel, layout, pattern, link, grid, cell, region, background |
| `text` | title, description, axis_label, label, annotation |
| `content` | number, data, domain, detail |
| `furniture` | legend, axes |
| `whole` | visualization |

New sub-categories may be introduced if clearly warranted.

````python
import json, sys, os
from pathlib import Path
from collections import Counter

import anthropic
from dotenv import load_dotenv

sys.path.insert(0, '../../vc_genome/scripts')
from _vc_canon import resolve_object_synset

load_dotenv(dotenv_path=Path('.').parent / '.env')

ROOT = Path('../..')
THREE_DIR = ROOT / 'vc_genome_output_full' / 'three_conditions'
OUT_FILE  = ROOT / 'vc_genome' / 'scripts' / '_vc_canon_extended_B.py'

with open(THREE_DIR / 'oar_B_510.json', encoding='utf-8') as f:
    b_raw = json.load(f)

print(f'Loaded {len(b_raw)} images')
````

## 1. Collect all unmapped names

````python
# Gather every object name that resolves to unknown.* (with corpus frequency)
raw_ctr = Counter()
for img, ext in b_raw.items():
    for obj in ext['objects']:
        name = (obj.get('name') or '').strip().lower().replace(' ', '_').replace('-', '_')
        if not name:
            continue
        synset = resolve_object_synset(name)
        if synset.startswith('unknown.'):
            raw_ctr[name] += 1

unknown_names = sorted(raw_ctr, key=lambda x: -raw_ctr[x])
print(f'Distinct unmapped names: {len(unknown_names)}')
print(f'Total unmapped tokens:   {sum(raw_ctr.values())}')
print()
print('Top 40 by frequency:')
for name in unknown_names[:40]:
    print(f'  {raw_ctr[name]:4d}  {name}')
````

## 2. Ask Claude to assign synsets

````python
SYSTEM_PROMPT = """You are a visualization taxonomy expert.
You will receive a list of object names extracted from participant comments about data visualizations.
Your job is to assign each name to a synset from the taxonomy below.

## Taxonomy
Each synset is written as `category.subcategory`. Use the closest match.
You MAY introduce a new subcategory (not listed) if clearly warranted, but prefer existing ones.

mark.bar          – bar marks in a bar chart
mark.point        – points, dots, circles, scatter marks
mark.line         – lines, curves, strokes
mark.color_segment – filled color bands or regions used as marks
mark.glyph        – composite or custom glyph marks
mark.node         – nodes in a network/tree diagram
mark.area         – filled area marks
mark.shape        – generic geometric shapes
mark.symbol       – icons, arrows, symbols
mark.element      – generic visual element (fallback mark)

property.color    – color, hue, saturation, palette, shade, tint
property.texture  – texture, pattern fill, hatching, stippling (NOT color)
property.font_size – font size, text size only (e.g. "font_size", "small_font")
property.font_style – font face, typeface, font weight, typography, font styling, font family (broader than size)
property.encoding – visual channel / data encoding
property.size     – size or scale of marks
property.opacity  – transparency or opacity

structure.panel   – bounding box, panel, container
structure.layout  – spatial arrangement / layout
structure.pattern – repeated visual pattern
structure.link    – edges, connections, links
structure.grid    – grid structure, matrix layout
structure.cell    – individual grid cell
structure.region  – geographic or data region
structure.background – background layer
structure.hierarchy – tree or hierarchical structure (use if needed)

text.title        – chart or section title
text.description  – caption, subtitle, description
text.axis_label   – axis label or tick label
text.label        – data label, inline text, word-cloud word
text.annotation   – annotation, callout text
text.tooltip      – tooltip or hover text (use if needed)

content.number    – numeric values, quantities
content.data      – data points or dataset reference
content.domain    – domain knowledge, concepts
content.detail    – level of detail
content.category  – categorical variable or group label (use if needed)

furniture.legend  – legend, color key
furniture.axes    – axis lines, tick marks, gridlines
furniture.scale   – scale bar or reference scale (use if needed)

whole.visualization – the entire chart or visualization

## Instructions
Return ONLY a valid JSON object mapping each input name (string key) to its synset (string value).
No markdown fences, no explanation, no prose.
Example: {"trend_line": "mark.line", "hover_tooltip": "text.tooltip"}
"""

# Split into chunks of 80 names to stay within token limits
CHUNK_SIZE = 80
chunks = [unknown_names[i:i+CHUNK_SIZE] for i in range(0, len(unknown_names), CHUNK_SIZE)]
print(f'Will send {len(chunks)} API call(s) covering {len(unknown_names)} names')
````

````python
api_key = os.environ.get('ANTHROPIC_API_KEY')
if not api_key:
    raise RuntimeError('ANTHROPIC_API_KEY not set')

client = anthropic.Anthropic(api_key=api_key)

extended_map = {}   # raw_name -> synset
errors       = []

for idx, chunk in enumerate(chunks):
    user_msg = json.dumps(chunk, ensure_ascii=False)
    resp = client.messages.create(
        model='claude-opus-4-6',
        max_tokens=2048,
        temperature=0.0,
        system=SYSTEM_PROMPT,
        messages=[{'role': 'user', 'content': user_msg}],
    )
    raw = resp.content[0].text.strip()
    # Strip markdown fences if model adds them
    if raw.startswith('```'):
        raw = raw.split('```')[1]
        if raw.startswith('json'):
            raw = raw[4:]
    try:
        parsed = json.loads(raw)
        extended_map.update(parsed)
        print(f'Chunk {idx+1}/{len(chunks)}: got {len(parsed)} mappings')
    except json.JSONDecodeError as e:
        print(f'Chunk {idx+1} parse error: {e}\nRaw output:\n{raw[:300]}')
        errors.append((idx, raw))

print(f'\nTotal mappings received: {len(extended_map)}')
print(f'Errors: {len(errors)}')
````

## 3. Inspect a sample of mappings

````python
import pandas as pd

sample_df = pd.DataFrame([
    {'raw_name': k, 'synset': v, 'freq': raw_ctr[k]}
    for k, v in extended_map.items()
]).sort_values('freq', ascending=False)

print(f'Category distribution:')
cats = sample_df['synset'].str.split('.').str[0].value_counts()
print(cats.to_string())
print()
display(sample_df.head(50))
````

## 4. Save as `scripts/_vc_canon_extended_B.py`

This file is a drop-in supplement covering the **Human-B condition** (`oar_B_510.json`).
Import it alongside `_vc_canon.py` and use `resolve_object_synset_extended()` which tries
the extended map first before falling back to the original resolver.

> **Note:** LLM-V1 and LLM-V2 conditions have additional unmapped names not covered here.
> A separate `_vc_canon_extended_V.py` should be built from those files if needed.


````python
# Validate: every assigned synset should follow category.subcategory format
bad = {k: v for k, v in extended_map.items() if '.' not in v or v.startswith('unknown')}
if bad:
    print(f'WARNING: {len(bad)} entries have malformed synsets:')
    for k, v in list(bad.items())[:20]:
        print(f'  {k!r} -> {v!r}')
else:
    print('All synsets well-formed.')

# Sort by synset then name for readable output
sorted_map = dict(sorted(extended_map.items(), key=lambda x: (x[1], x[0])))

# Group by category for annotated output
from collections import defaultdict
by_cat = defaultdict(dict)
for name, syn in sorted_map.items():
    by_cat[syn.split('.')[0]][name] = syn

# ── Build file content ─────────────────────────────────────────────────────
header = (
    '"""\n'
    'Extended object synset mappings — generated by BuildExpandedSynset.ipynb.\n'
    'Maps previously-unmapped (unknown.*) object names from oar_B_510.json\n'
    'to the canonical category.subcategory taxonomy.\n'
    '\n'
    'NOTE: This file covers the Human-B condition only (oar_B_510.json).\n'
    '      LLM-V1 and LLM-V2 conditions have additional unmapped names not covered here.\n'
    '\n'
    'DO NOT overwrite _vc_canon.py. Import this alongside it:\n'
    '\n'
    '    from _vc_canon import resolve_object_synset\n'
    '    from _vc_canon_extended_B import resolve_object_synset_extended\n'
    '"""\n'
    'from __future__ import annotations\n'
    'from _vc_canon import resolve_object_synset as _base_resolve\n'
    '\n'
    'EXTENDED_SYNSETS: dict[str, str] = {\n'
)

dict_lines = []
for cat in sorted(by_cat):
    dict_lines.append(f'    # {cat}')
    for name, syn in sorted(by_cat[cat].items(), key=lambda x: x[1]):
        dict_lines.append(f'    {name!r}: {syn!r},')

footer = (
    '}\n'
    '\n'
    '\n'
    'def resolve_object_synset_extended(name: str) -> str:\n'
    '    """Like resolve_object_synset but also checks EXTENDED_SYNSETS."""\n'
    '    base = _base_resolve(name)\n'
    '    if not base.startswith("unknown."):\n'
    '        return base\n'
    '    # Normalise the same way _vc_canon does\n'
    '    norm = (name or "").strip().lower().replace(" ", "_").replace("-", "_")\n'
    '    return EXTENDED_SYNSETS.get(norm, base)\n'
)

content = header + '\n'.join(dict_lines) + '\n' + footer
OUT_FILE.write_text(content, encoding='utf-8')
print(f'Saved {len(extended_map)} entries → {OUT_FILE}')
print(f'\nPreview of function:')
print(footer)
````

## 5. Quick verification

````python
import importlib, sys

# Force reload in case the module was already imported
if '_vc_canon_extended_B' in sys.modules:
    del sys.modules['_vc_canon_extended_B']

from _vc_canon_extended_B import resolve_object_synset_extended

# Re-count unknowns using the extended resolver
still_unknown = Counter()
for img, ext in b_raw.items():
    for obj in ext['objects']:
        syn = resolve_object_synset_extended(obj.get('name', ''))
        if syn.startswith('unknown.'):
            still_unknown[syn] += 1

orig_unknown_total = sum(raw_ctr.values())
remaining = sum(still_unknown.values())
resolved  = orig_unknown_total - remaining

print(f'Originally unmapped tokens: {orig_unknown_total}')
print(f'Resolved by extended map:   {resolved}  ({100*resolved/orig_unknown_total:.1f}%)')
print(f'Still unknown:              {remaining}')
if still_unknown:
    print('\nTop remaining unknowns:')
    for syn, cnt in still_unknown.most_common(20):
        print(f'  {cnt:4d}  {syn}')
````

## Synset Lookup Table
Combined view of all raw-name → synset mappings: original handcrafted entries + Claude-generated extended entries.

````python
import sys, pandas as pd
from pathlib import Path

sys.path.insert(0, '../../vc_genome/scripts')
from _vc_canon import OBJECT_SYNSETS
from _vc_canon_extended_B import EXTENDED_SYNSETS

# Combine: original entries first, then extended (Claude-assigned)
# For names that appear in both, extended takes precedence
all_mappings = []
for raw_name, synset in sorted(OBJECT_SYNSETS.items()):
    all_mappings.append({
        'raw_name': raw_name,
        'synset':   synset,
        'category': synset.split('.')[0] if '.' in synset else synset,
        'source':   'original',
    })
for raw_name, synset in sorted(EXTENDED_SYNSETS.items()):
    all_mappings.append({
        'raw_name': raw_name,
        'synset':   synset,
        'category': synset.split('.')[0] if '.' in synset else synset,
        'source':   'extended',
    })

synset_df = pd.DataFrame(all_mappings)

OUT = Path('vc_genome_output_full/vistype_profile')
OUT.mkdir(parents=True, exist_ok=True)
synset_df.to_csv(OUT / 'synset_mappings_B.csv', index=False)
print(f'Total mappings: {len(synset_df)}  '
      f'(original: {(synset_df.source=="original").sum()}, '
      f'extended: {(synset_df.source=="extended").sum()})')
display(synset_df)
````

````python
# ── Unique canonical synsets across both dictionaries ─────────────────────
unique_synsets = sorted(set(OBJECT_SYNSETS.values()) | set(EXTENDED_SYNSETS.values()))

synset_list_df = pd.DataFrame([
    {
        'synset':   s,
        'category': s.split('.')[0] if '.' in s else s,
        'n_raw_names': sum(1 for v in {**OBJECT_SYNSETS, **EXTENDED_SYNSETS}.values() if v == s),
    }
    for s in unique_synsets
]).sort_values(['category', 'synset'])

print(f'Unique canonical synsets: {len(synset_list_df)}')
print(f'Categories: {synset_list_df.category.nunique()}')
display(synset_list_df)
````
